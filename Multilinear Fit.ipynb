```python
import numpy as np
from sklearn.linear_model import LinearRegression

# Adapted from:
# https://www.datasciencecentral.com/profiles/blogs/linear-regression-in-python-use-of-numpy-scipy-and-statsmodels,
# https://realpython.com/linear-regression-in-python/#multiple-linear-regression

# Goal: Find logMdot = A + blogL + dlogP
# Goal: Find logMdot = A + blogL + clogM
# Goal: Find logMdot = A + blogL + clogM + dlogP
```

```python
DataToFit = np.transpose(np.asarray(np.genfromtxt('ObservablePlus_FundMiraData.csv',delimiter=',')))
```

```python
def ErrorClean (Data_Set,Error_limit_Mdot,Error_limit_L): # A function to clear all data with unreliable values in Mdot or L
    CuttingArray = []
    i = 0
    Cleaned_Set = np.transpose(Data_Set)
    while i<len(Data_Set[3]):
        if Data_Set[7][i] > Error_limit_Mdot: # If the relative error is greater than specified limit for the dust mass-loss rate
            CuttingArray.append(i)
        if Data_Set[10][i] > Error_limit_L: # If the relative error is greater than specified limit for the luminosity
            CuttingArray.append(i)
        i += 1
    Cleaned_Set = np.delete(Cleaned_Set,CuttingArray,axis=0)
    print ('Entries Removed:',len(CuttingArray))
    print ('Stars Remaining:', len(Cleaned_Set))
    return np.transpose(Cleaned_Set)
        
```

```python
DataToFit = ErrorClean(DataToFit,1,0.1)
```

```python
logMdotTotal = np.log10(DataToFit[5])
```

```python
logL = np.log10(DataToFit[8])
logP = np.log10(DataToFit[13])
logM = np.log10(DataToFit[18])
```

```python
print(logL,logP,logM)
```

```python
inArray = np.transpose(np.vstack((logL,logP))) # First, let's find logMdot = A + blogL + dlogP
x = inArray
y = logMdotTotal
# x, y = np.array(x), np.array(y)
model = LinearRegression().fit(x, y)
# Returns a linear regression y = b0 + sum(b_i*x_i,i=1 to n)
r_sq = model.score(x, y)
print('coefficient of determination:', r_sq)
print('intercept:', model.intercept_) # b0
print('slope:', model.coef_) # b1, b2, b3,..., bn
```

This result tells us that the multilinear fit to this data suggests logMdot = -7.60 - 0.8538\*logL +1.738\*logP

```python
inArray = np.transpose(np.vstack((logL,logM))) # Next, let's find logMdot = A + blogL + clogM
x = inArray
y = logMdotTotal
# x, y = np.array(x), np.array(y)
model = LinearRegression().fit(x, y)
# Returns a linear regression y = b0 + sum(b_i*x_i,i=1 to n)
r_sq = model.score(x, y)
print('coefficient of determination:', r_sq)
print('intercept:', model.intercept_) # b0
print('slope:', model.coef_) # b1, b2, b3,..., bn
```

This result tells us that the multilinear fit to this data suggests logMdot = -11.48 + 1.345\*logL - 1.492\*logP
