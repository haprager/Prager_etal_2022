```python
def Runge_Kutta_Order_Four (f, a, b, N, alpha): # Requires 5 inputs: a
    # mathematical function "f", a starting limit "a", an ending limit "b",
    # the number of steps "N", and an initial value "alpha"
    # Step 1: Initialize variables that will be used in the numerical
    # approximation
    h = (b-a)/float(N)
    t = np.zeros(N+1)
    t[0] = a
    w = np.zeros(N+1)
    w[0] = alpha
    # Step 2: Set-up a for loop to perform the approximation
    for i in range (1,N+1):
        # Step 3: Establish values for nested variables
        K1 = h*f(t[i-1],w[i-1])
        K2 = h*f(t[i-1]+h/2., w[i-1]+K1/2.)
        K3 = h*f(t[i-1]+h/2., w[i-1]+K2/2.)
        K4 = h*f(t[i-1]+h, w[i-1]+K3)
        
        # Step 4: Perform a Runge-Kutta approximation of order 4 at
        # w_i
        w[i] = w[i-1]+(K1+2*K2+2*K3+K4)/6.
        t[i] = a + i*h
        
    # Step 5: Create an array of w and t for output
    OutputArray = np.array([t, w])
    return OutputArray
```

```python
def Runge_Kutta_Order_Four_Sys ( fList, a, b, N, alphaList ):
    '''This function implements the fourth order Runge-Kutta method. It takes 5 inputs:
    a system of 1st order differential equations fList, 
    a start point for the independent variable a,
    an endpoint for the independent variable b, 
    the number of equally spaced intervals between a and b N, 
    and a list of initial conditions for the diff. eqns. alphaList. 
    The function will return aprroximations to the input functions at 
    the (N+1) values of the independent variable.'''
    '''Note: fList should be input such that it takes two inputs: an independent var. and a list of dependent vars.'''
    m = len(fList) # First, get the order of the system
    print('Order of system:', m)
    h = (b-a)/float(N) # Calculate the size of each step h; float(N) shouldn't be necessary in Python 3, but it does make backporting easier
    t = np.zeros(N+1) # Create a list to hold the current value of the independent variable
    t[0] = a # Set the first value to the startpoint a
    w = np.zeros((N+1, int(m))) # Create an empty array to hold all of the values for the solution at each t
    for j in range (1, m+1): # Assign the intial values of alphaList to the first row of w
        w[0][j-1] = alphaList[j-1]
    k1 = np.zeros(m) # Initialize a list to hold k_1 values
    k2 = np.zeros(m) # ... k_2 values
    k3 = np.zeros(m) # ... k_3 values
    k4 = np.zeros(m) # ... k_4 values
    for i in range(1, N+1): # Perform the method at each step until N is reached
        for j in range (1,m+1):
            fj = fList[j-1] # Set the current function to be worked on to be the jth entry of fList
            k1[j-1] = h*fj(t[i-1],w[i-1]) # Calculate k1 for the jth diff. eq.
        wk2 = w[i-1]+k1/2
        for j in range (1,m+1):
            fj = fList[j-1]
            k2[j-1] = h* fj(t[i-1]+h/2.,wk2)
        wk3 = w[i-1]+(1/2)*k2
        for j in range(1,m+1):
            fj = fList[j-1]
            k3[j-1] = h* fj(t[i-1]+h/2.,wk3)
        wk4 = w[i-1]+k3
        for j in range(1,m+1):
            fj = fList[j-1]
            k4[j-1] = h* fj(t[i-1]+h/2.,wk4)
        for j in range(1,m+1):
            w[i][j-1] = w[i-1][j-1]+(k1[j-1]+2*k2[j-1]+2*k3[j-1]+k4[j-1])/6.
        t[i] = a + i*h
        #print("Loop", i, "completed.", float(i)/float(N)*100, "percent complete,")
    concatenatable_t_array = np.zeros((1,N+1))
    for i in range (0,N+1):
        concatenatable_t_array[0][i-1] = t[i-1]
    #print(len(concatenatable_t_array[0]) == len(w.transpose()[0]))
    #print(concatenatable_t_array.transpose(),"\n", w)
    output_loop = np.concatenate((concatenatable_t_array.transpose(),w),axis = 1)
    return output_loop
```
